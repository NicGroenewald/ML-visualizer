# mlviz — Decision Tree Demo

Train any `DecisionTreeClassifier` and call `visualize()` to open an interactive browser view.

---

```python
from sklearn.tree import DecisionTreeClassifier
from mlviz import visualize
```

## Iris (3-class, 4 features)

A classic. The query point is the first sample in the dataset — watch the path bar at the bottom of the visualiser show exactly how the tree classified it.

```python
from sklearn.datasets import load_iris

iris = load_iris()
X, y = iris.data, iris.target

model = DecisionTreeClassifier(max_depth=3, random_state=0)
model.fit(X, y)

visualize(model, X, y, query=X[0], feature_names=iris.feature_names)
```

---

## Wine (3-class, 13 features)

```python
from sklearn.datasets import load_wine

wine = load_wine()
X, y = wine.data, wine.target

model = DecisionTreeClassifier(max_depth=4, random_state=0)
model.fit(X, y)

visualize(model, X, y, query=X[42], feature_names=wine.feature_names)
```

## Breast Cancer (2-class, 30 features)

```python
from sklearn.datasets import load_breast_cancer

bc = load_breast_cancer()
X, y = bc.data, bc.target

model = DecisionTreeClassifier(max_depth=4, random_state=0)
model.fit(X, y)

visualize(model, X, y, query=X[0], feature_names=bc.feature_names)
```
